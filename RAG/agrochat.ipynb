```python
import os
import json
import pandas as pd
from dotenv import load_dotenv
import kagglehub

# 1. Load environment variables from .env
load_dotenv(override=True)

username = os.getenv("KAGGLE_USERNAME")
key = os.getenv("KAGGLE_KEY")

if not username or not key:
    raise ValueError("KAGGLE_USERNAME or KAGGLE_KEY is missing from your .env file!")

# 2. Write the credentials to ~/.kaggle/kaggle.json directly
kaggle_dir = os.path.expanduser("~/.kaggle")
os.makedirs(kaggle_dir, exist_ok=True)

kaggle_json_path = os.path.join(kaggle_dir, "kaggle.json")
with open(kaggle_json_path, "w") as f:
    json.dump({"username": username, "key": key}, f)

# Set strictly required permissions for Kaggle API
os.chmod(kaggle_json_path, 0o600)

print(f"✅ Created {kaggle_json_path} for user: {username}")

```
